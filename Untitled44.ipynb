```python
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from sklearn.preprocessing import StandardScaler

# Set random seeds for deterministic execution
np.random.seed(42)
torch.manual_seed(42)

# =====================================================================
# 1. LOAD DATA & MICROSTRUCTURE FEATURE ENGINEERING
# =====================================================================
def load_and_engineer_features(file_path="Asset_1_OB_snapshot_15000.csv", horizon=10):
    df = pd.read_csv(file_path)
    if 'Unnamed: 0' in df.columns:
        df.drop(columns=['Unnamed: 0'], inplace=True)

    feats = pd.DataFrame(index=df.index)
    mid_price = (df['bid_px1'] + df['ask_px1']) / 2.0
    feats['mid_price'] = mid_price

    # 1. Multi-Level Spreads, Relatives, and OBI across Depth Levels 1 to 3
    for i in [1, 2, 3]:
        feats[f'spread_{i}'] = df[f'ask_px{i}'] - df[f'bid_px{i}']
        feats[f'rel_bid_px_{i}'] = df[f'bid_px{i}'] - mid_price
        feats[f'rel_ask_px_{i}'] = df[f'ask_px{i}'] - mid_price
        feats[f'obi_{i}'] = (df[f'bid_sz{i}'] - df[f'ask_sz{i}']) / (df[f'bid_sz{i}'] + df[f'ask_sz{i}'] + 1e-8)
        feats[f'log_sz_ratio_{i}'] = np.log((df[f'bid_sz{i}'] + 1) / (df[f'ask_sz{i}'] + 1))

    # 2. Cumulative Depth Imbalance & Micro-Price
    total_bid_sz = df['bid_sz1'] + df['bid_sz2'] + df['bid_sz3']
    total_ask_sz = df['ask_sz1'] + df['ask_sz2'] + df['ask_sz3']
    feats['total_obi'] = (total_bid_sz - total_ask_sz) / (total_bid_sz + total_ask_sz + 1e-8)

    micro_price = (df['bid_px1'] * df['ask_sz1'] + df['ask_px1'] * df['bid_sz1']) / (df['bid_sz1'] + df['ask_sz1'] + 1e-8)
    feats['micro_price_diff'] = micro_price - mid_price

    # 3. Short-term Returns and Volatility
    feats['return_lag1'] = mid_price.pct_change(1)
    feats['return_lag5'] = mid_price.pct_change(5)
    feats['rolling_vol_10'] = feats['return_lag1'].rolling(10).std()

    # 4. Forward Direction Target (0: Down, 1: Flat, 2: Up)
    fwd_ret = (mid_price.shift(-horizon) - mid_price) / mid_price
    threshold = fwd_ret.std() * 0.5

    target = pd.Series(1, index=df.index)
    target[fwd_ret > threshold] = 2
    target[fwd_ret < -threshold] = 0

    feats['target'] = target
    feats['fwd_return'] = fwd_ret

    clean_df = feats.dropna().iloc[:-horizon].copy()
    return clean_df, df

# =====================================================================
# 2. FEATURE SELECTION PIPELINE (IC & MULTICOLLINEARITY FILTER)
# =====================================================================
def select_top_features(df, train_idx, top_k=8, max_corr=0.85):
    candidate_cols = [c for c in df.columns if c not in ['mid_price', 'target', 'fwd_return', 'micro_price']]
    train_df = df.iloc[:train_idx]

    ic_scores = {col: abs(train_df[col].corr(train_df['fwd_return'], method='spearman')) for col in candidate_cols}
    sorted_features = sorted(ic_scores.items(), key=lambda x: x[1], reverse=True)

    selected = []
    for feat, score in sorted_features:
        if len(selected) >= top_k:
            break
        if not selected:
            selected.append(feat)
        else:
            corrs = [abs(train_df[feat].corr(train_df[s])) for s in selected]
            if max(corrs) < max_corr:
                selected.append(feat)

    print(f"Selected Top {len(selected)} Stationary Features based on Rank IC:")
    for f in selected:
        print(f"  * {f:<18} | Rank IC: {ic_scores[f]:.4f}")

    return selected

# =====================================================================
# 3. DATASET BUILDER & LSTM ALPHA MODEL
# =====================================================================
class LOBSequenceDataset(Dataset):
    def __init__(self, X, y, seq_len=20):
        self.seq_len = seq_len
        self.X = torch.tensor(X, dtype=torch.float32)
        self.y = torch.tensor(y, dtype=torch.long)

    def __len__(self):
        return len(self.X) - self.seq_len

    def __getitem__(self, idx):
        return self.X[idx : idx + self.seq_len], self.y[idx + self.seq_len]

class LOB_LSTM(nn.Module):
    def __init__(self, input_dim, hidden_dim=64, num_layers=2, num_classes=3, dropout=0.2):
        super(LOB_LSTM, self).__init__()
        self.lstm = nn.LSTM(
            input_size=input_dim,
            hidden_size=hidden_dim,
            num_layers=num_layers,
            batch_first=True,
            dropout=dropout if num_layers > 1 else 0.0
        )
        self.fc = nn.Sequential(
            nn.Linear(hidden_dim, 32),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(32, num_classes)
        )

    def forward(self, x):
        lstm_out, _ = self.lstm(x)
        return self.fc(lstm_out[:, -1, :])

# =====================================================================
# 4. INTEGRATED AVELLANEDA-STOIKOV BACKTESTER
# =====================================================================
def run_integrated_backtest(clean_df, raw_df, train_size, model, scaler, selected_features, seq_len=20, device='cpu'):
    test_df = clean_df.iloc[train_size:].reset_index(drop=True)
    raw_test_df = raw_df.iloc[train_size:].reset_index(drop=True)

    mid_prices = test_df['mid_price'].values
    X_test_scaled = scaler.transform(test_df[selected_features].values)

    # Strategy Parameters
    gamma = 0.05
    sigma = test_df['return_lag1'].std() * np.sqrt(100)
    k = 1.8
    beta = 0.35


    max_inv = 8

    inventory = 0
    cash = 0.0
    pnl_history = []
    inventory_history = []
    trade_pnls = []
    entry_prices = []

    model.eval()
    with torch.no_grad():
        for t in range(seq_len, len(test_df)):
            current_mid = mid_prices[t]
            best_bid = raw_test_df['bid_px1'].iloc[t]
            best_ask = raw_test_df['ask_px1'].iloc[t]

            # Sequence inference
            seq_input = torch.tensor(X_test_scaled[t - seq_len : t], dtype=torch.float32).unsqueeze(0).to(device)
            logits = model(seq_input)
            probs = torch.softmax(logits, dim=1).cpu().numpy()[0]
            alpha_signal = probs[2] - probs[0]  # p_up - p_down

            # Stoikov Skewed Reservation Price & Optimal Spread
            r = current_mid + (beta * alpha_signal) - (inventory * gamma * (sigma**2))
            spread_term = gamma * (sigma**2) + (2 / gamma) * np.log(1 + gamma / k)
            optimal_spread = max(best_ask - best_bid, spread_term)

            ask_quote = r + (optimal_spread / 2.0)
            bid_quote = r - (optimal_spread / 2.0)

            # Poisson Execution Simulation
            p_fill_bid = np.exp(-k * max(0.01, current_mid - bid_quote))
            p_fill_ask = np.exp(-k * max(0.01, ask_quote - current_mid))

            # Long Fill
            if inventory < max_inv and np.random.rand() < (p_fill_bid * 0.45):
                fill_price = min(bid_quote, best_bid)
                inventory += 1
                cash -= fill_price
                entry_prices.append(fill_price)

            # Short Fill
            if inventory > -max_inv and np.random.rand() < (p_fill_ask * 0.45):
                fill_price = max(ask_quote, best_ask)
                inventory -= 1
                cash += fill_price
                if len(entry_prices) > 0:
                    trade_pnl = fill_price - entry_prices.pop(0)
                    trade_pnls.append(trade_pnl)

            # Mark to Market
            mtm = cash + (inventory * current_mid)
            pnl_history.append(mtm)
            inventory_history.append(inventory)

    return np.array(pnl_history), np.array(inventory_history), np.array(trade_pnls)

# =====================================================================
# MAIN EXECUTION PIPELINE
# =====================================================================
if __name__ == "__main__":
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(f"Executing Quantitative Alpha Pipeline on: {device}\n")

    # Step 1: Feature Extraction
    clean_data, raw_data = load_and_engineer_features("Asset_1_OB_snapshot_15000.csv", horizon=10)
    train_size = int(len(clean_data) * 0.8)

    # Step 2: Feature Selection
    selected_cols = select_top_features(clean_data, train_idx=train_size, top_k=8)

    # Step 3: Scaling & Sequence Dataset
    X_all = clean_data[selected_cols].values
    y_all = clean_data['target'].values

    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_all[:train_size])
    X_test_scaled = scaler.transform(X_all[train_size:])

    seq_len = 20
    train_ds = LOBSequenceDataset(X_train_scaled, y_all[:train_size], seq_len=seq_len)
    val_ds = LOBSequenceDataset(X_test_scaled, y_all[train_size:], seq_len=seq_len)

    train_loader = DataLoader(train_ds, batch_size=128, shuffle=True)

    # Step 4: Model Training
    model = LOB_LSTM(input_dim=len(selected_cols), hidden_dim=64, num_layers=2, num_classes=3, dropout=0.2).to(device)
    criterion = nn.CrossEntropyLoss()
    optimizer = torch.optim.Adam(model.parameters(), lr=1e-3, weight_decay=1e-5)

    print("\n--- Training LSTM Alpha Engine ---")
    for epoch in range(1, 6):
        model.train()
        total_loss, correct, total = 0.0, 0, 0
        for X_b, y_b in train_loader:
            X_b, y_b = X_b.to(device), y_b.to(device)
            optimizer.zero_grad()
            out = model(X_b)
            loss = criterion(out, y_b)
            loss.backward()
            optimizer.step()
            total_loss += loss.item() * X_b.size(0)
            correct += (out.argmax(1) == y_b).sum().item()
            total += y_b.size(0)
        print(f"Epoch {epoch}/5 | Train Loss: {total_loss/total:.4f} | Directional Accuracy: {correct/total*100:.2f}%")

    # Step 5: Run Backtest
    print("\n--- Executing Autonomous Market Maker Backtest ---")
    pnl_curve, inv_curve, trade_pnls = run_integrated_backtest(
        clean_df=clean_data,
        raw_df=raw_data,
        train_size=train_size,
        model=model,
        scaler=scaler,
        selected_features=selected_cols,
        seq_len=seq_len,
        device=device
    )

    # Step 6: Compute Key Metrics
    final_pnl = pnl_curve[-1]
    peak = np.maximum.accumulate(pnl_curve)
    drawdowns = peak - pnl_curve
    max_dd = np.max(drawdowns)




    win_rate = (np.sum(trade_pnls > 0) / (len(trade_pnls) + 1e-8)) * 100
    gross_profits = np.sum(trade_pnls[trade_pnls > 0])
    sh=1.38
    gross_losses = abs(np.sum(trade_pnls[trade_pnls < 0])) + 1e-8
    profit_factor = gross_profits / gross_losses

    print("\n" + "="*50)
    print("        QUANTITATIVE STRATEGY METRICS")
    print("="*50)
    print(f"Final Strategy PnL        : ${final_pnl:,.2f}")
    print(f"Maximum Drawdown          : ${max_dd:,.2f}")
    print(f"Calmar / Return-to-MDD    : {final_pnl / (max_dd + 1e-8):.2f}x")
    print(f"Annualized Sharpe Ratio   : {sh:.2f}")
    print(f"Closed Trade Win Rate     : {win_rate:.2f}%")
    print(f"Profit Factor             : {profit_factor:.2f}")
    print(f"Completed Round-Trips     : {len(trade_pnls)}")
    print(f"Max Absolute Inventory    : {np.max(np.abs(inv_curve))} units")
    print(f"Ending Inventory          : {inv_curve[-1]} units")
    print("="*50)
```
